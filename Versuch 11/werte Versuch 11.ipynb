```python
import numpy as np 


def auswertung(messwerte, name):
    unsicherheiten = np.full(messwerte.shape, 0.4, dtype=float)
    werte = messwerte / 3
    werte_unsicherheiten = unsicherheiten / 3
    mittelwert = np.mean(werte)
    mittelwert_unsicherheit = np.sqrt(np.sum(werte_unsicherheiten**2)) / len(werte)

    print(werte)
    print(f"{name}: {mittelwert:.6f} +/- {mittelwert_unsicherheit:.6f}")
    print(f"Streuung: {np.std(werte):.6f}")
```

Maximaldurchgang

```python

a = np.array([4.96, 5.01, 4.98, 5.01, 5.03, 5.1, 4.89, 5.06, 5.01, 5.09])
print (a/3)
print (np.mean(a/3))
print(np.std(a/3))

auswertung(a, "a")
```

Nulldurchgang

```python

b = np.array([5.12, 5.04, 5.01, 5.00, 4.96, 5.00, 5.08, 5.00, 5.03, 5.01])

print (b/3)
print (np.mean(b/3))
print(np.std(b/3))


auswertung(b, "b") 
```
