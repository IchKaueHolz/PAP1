Aufgabe 1: Mittelwert aus den Abständen des Quincke'schen Rohr bilden

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Absolute Höhen h in cm 
error = 0.05

luft = np.array([99.2, 90.6, 82.1, 73.6, 64.8, 56.0, 47.9, 39.0, 30.4, 21.5, 12.7])
diffluft = np.diff(luft)

CO2 = np.array([98.4, 91.4, 84.5, 77.6, 70.8, 64.7, 57.6, 50.1, 44.0, 37.3, 30.9, 23.6, 16.9, 10.0])
diffCO2 = np.diff(CO2)
print(f'{diffluft=}')
print(f'{diffCO2=}')

LuftMean = np.mean(diffluft)
LuftStd = np.std(diffluft, ddof=1)
print(f'{LuftMean=:.3f},{LuftStd=:.3f}')

CO2Mean = np.mean(diffCO2)
CO2Std = np.std(diffCO2, ddof=1)
print(f'{CO2Mean=:.3f},{CO2Std=:.3f}')

print(f'Fehler Luft: {np.sqrt(LuftStd**2 + error**2):.3f}')
print(f'Fehler CO2: {np.sqrt(CO2Std**2 + error**2):.3f}')

```

Literaturwert C

```python
def calc_c(k, M):
    R = 8.31446261815324 #allgemeine Gaskonstante in J/(mol*K)
    T = 273.15 #Temperatur in K√
    return np.sqrt(k * R * T / M)

print(f'c_Luft = {calc_c(1.4, 0.029):.3f} m/s')
print(f'c_CO2 = {calc_c(1.3, 0.044):.3f} m/s')
```

Laufzeitmessung

```python
import numpy as np
from scipy.optimize import curve_fit

# 1. Daten aus dem Protokoll eintragen (Einheit: cm)
d_norm = np.array([6.0, 9.4, 12.9, 16.4, 20.0, 23.5, 26.9])
u_d_norm = np.full_like(d_norm, 0.3)
n_norm = np.arange(len(d_norm)) # 0, 1, 2, ...

d_liss = np.array([4.2, 5.9, 7.6, 9.4, 11.2, 12.9, 14.7, 16.5, 18.2, 19.9, 21.7, 23.5, 25.2, 26.9])
u_d_liss = np.full_like(d_liss, 0.2)
n_liss = np.arange(len(d_liss)) # 0, 1, 2, ...

f = 10000 # Hz (10 kHz)

# 2. Lineare Fit-Funktion definieren
def linear(x, a, b):
    return a * x + b

# 3. Fit für Normalmessung (Steigung a = lambda)
popt_n, pcov_n = curve_fit(linear, n_norm, d_norm, sigma=u_d_norm, absolute_sigma=True)
lambda_norm = popt_n[0]
u_lambda_norm = np.sqrt(np.diag(pcov_n))[0]

# 4. Fit für Lissajous-Messung (Steigung a = lambda / 2)
popt_l, pcov_l = curve_fit(linear, n_liss, d_liss, sigma=u_d_liss, absolute_sigma=True)
lambda_halb_liss = popt_l[0]
u_lambda_halb_liss = np.sqrt(np.diag(pcov_l))[0]

# Lissajous Werte verdoppeln für lambda
lambda_liss = 2 * lambda_halb_liss
u_lambda_liss = 2 * u_lambda_halb_liss

# 5. Schallgeschwindigkeit berechnen (c = lambda * f), Umrechnung cm -> m
c_norm = lambda_norm * 0.01 * f
u_c_norm = u_lambda_norm * 0.01 * f

c_liss = lambda_liss * 0.01 * f
u_c_liss = u_lambda_liss * 0.01 * f

# 6. Ergebnisse ausgeben
print("--- Ergebnisse Normalmessung ---")
print(f"Lambda = ({lambda_norm:.3f} +/- {u_lambda_norm:.3f}) cm")
print(f"c      = ({c_norm:.1f} +/- {u_c_norm:.1f}) m/s\n")

print("--- Ergebnisse Lissajous-Messung ---")
print(f"Lambda = ({lambda_liss:.3f} +/- {u_lambda_liss:.3f}) cm")
print(f"c      = ({c_liss:.1f} +/- {u_c_liss:.1f}) m/s")
```

```python
import numpy as np
from scipy.optimize import curve_fit

# norm= 2khz
d_norm = np.array([11.3, 20.1])
u_d_norm = np.full_like(d_norm, 0.2)
n_norm = np.arange(len(d_norm)) # 0, 1, 2, ...

# liss= 5khz
d_liss = np.array([5.2, 8.3, 12.2, 15.1, 19.1, 22.6, 26.2])
u_d_liss = np.full_like(d_liss, 0.2)
n_liss = np.arange(len(d_liss)) # 0, 1, 2, ...


f1 = 2000 # Hz 
f2 = 5000 # Hz

def linear(x, a, b):
    return a * x + b

# 3. Fit für Normalmessung (Steigung a = lambda)
popt_n, pcov_n = curve_fit(linear, n_norm, d_norm, sigma=u_d_norm, absolute_sigma=True)
lambda_halb_norm = popt_n[0]
u_lambda_halb_norm = np.sqrt(np.diag(pcov_n))[0]

# 4. Fit für Lissajous-Messung (Steigung a = lambda / 2)
popt_l, pcov_l = curve_fit(linear, n_liss, d_liss, sigma=u_d_liss, absolute_sigma=True)
lambda_halb_liss = popt_l[0]
u_lambda_halb_liss = np.sqrt(np.diag(pcov_l))[0]

# Lissajous Werte verdoppeln da bbeide Messungen mit 180 Grad Phasenverschiebung
lambda_norm = 2 * lambda_halb_norm
u_lambda_norm = 2 * u_lambda_halb_norm

lambda_liss = 2 * lambda_halb_liss
u_lambda_liss = 2 * u_lambda_halb_liss

# 5. Schallgeschwindigkeit berechnen (c = lambda * f), Umrechnung cm -> m
c_norm = lambda_norm * 0.01 * f1
u_c_norm = u_lambda_norm * 0.01 * f1

c_liss = lambda_liss * 0.01 * f2
u_c_liss = u_lambda_liss * 0.01 * f2

# 6. Ergebnisse ausgeben
print("--- Ergebnisse Normalmessung ---")
print(f"Lambda = ({lambda_norm:.3f} +/- {u_lambda_norm:.3f}) cm")
print(f"c      = ({c_norm:.1f} +/- {u_c_norm:.1f}) m/s\n")

print("--- Ergebnisse Lissajous-Messung ---")
print(f"Lambda = ({lambda_liss:.3f} +/- {u_lambda_liss:.3f}) cm")
print(f"c      = ({c_liss:.1f} +/- {u_c_liss:.1f}) m/s")
```
