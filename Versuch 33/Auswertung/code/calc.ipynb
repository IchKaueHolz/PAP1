Rydberg berechnen

```python
import numpy as np
import uncertainties as unc
import uncertainties.unumpy as unp
def rydberg_constant(wavelength, n):
    m = 2 
    return 1 / wavelength * 1/(-1 / n**2 + 1 / m**2)

rot = unc.ufloat(666, 2) * 1e-9
tuerkis = unc.ufloat(494, 2) * 1e-9
violett = unc.ufloat(442, 2) * 1e-9
violett_dunkel = unc.ufloat(416, 2) * 1e-9

print("(n=3):", rydberg_constant(rot, 3))
print("(n=4):", rydberg_constant(tuerkis, 4))
print("(n=5):", rydberg_constant(violett, 5))
print("(n=6):", rydberg_constant(violett_dunkel, 6))

all = np.array([rydberg_constant(rot, 3).nominal_value, rydberg_constant(tuerkis, 4).nominal_value, rydberg_constant(violett, 5).nominal_value, rydberg_constant(violett_dunkel, 6).nominal_value])
print(f'Mean: {np.mean(all)}, Std: {np.std(all, ddof=1)}')
```
