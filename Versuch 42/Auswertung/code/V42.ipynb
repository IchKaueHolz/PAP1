```python
T_bar_graphit = 28.8
T_bar_alu = 34.4
T_bar_blei = 36.7

T_bar_graphit = unc.ufloat(T_bar_graphit, (T_bar_graphit*0.003+1)**2 + 0.1**2)
T_bar_alu = unc.ufloat(T_bar_alu, (T_bar_alu*0.003+1)**2 + 0.1**2)
T_bar_blei = unc.ufloat(T_bar_blei, (T_bar_blei*0.003+1)**2 + 0.1**2)

T_bar_graphit, T_bar_alu, T_bar_blei
```

```python
T1_graphit =98.6
T1_alu = 99.9
T1_blei = 100

T1_graphit = unc.ufloat(T1_graphit, (T1_graphit*0.003+1)**2 + 0.1**2)
T1_alu = unc.ufloat(T1_alu, (T1_alu*0.003+1)**2 + 0.1**2)
T1_blei = unc.ufloat(T1_blei, (T1_blei*0.003+1)**2 + 0.1**2)

T1_graphit, T1_alu, T1_blei
```

```python
T2_graphit = 24.8
T2_alu = 28.7
T2_blei = 33.8

T2_graphit = unc.ufloat(T2_graphit, (T2_graphit*0.003+1)**2 + 0.1**2)
T2_alu = unc.ufloat(T2_alu, (T2_alu*0.003+1)**2 + 0.1**2)
T2_blei = unc.ufloat(T2_blei, (T2_blei*0.003+1)**2 + 0.1**2)

T2_graphit, T2_alu, T2_blei

M_WK = unc.ufloat(571.80, 0.05)
```

```python
M_alu = unc.ufloat(155.70, 0.05)
M_graphit = unc.ufloat(125.1, 0.05)
M_blei = unc.ufloat(601.3, 0.05)

def c_x(T_bar, T1, T2, M_WK, M_K, M_x, c_w, W):
    return ((M_WK-M_K) * c_w + W) * (T_bar - T2) / (M_x * (T1 - T_bar))

c_graphit = c_x(T_bar_graphit, T1_graphit, T2_graphit, M_WK, M_K, M_graphit, c_w, W)
c_alu = c_x(T_bar_alu, T1_alu, T2_alu, M_WK, M_K, M_alu, c_w, W)
c_blei = c_x(T_bar_blei, T1_blei, T2_blei, M_WK, M_K, M_blei, c_w, W)

c_graphit, c_alu, c_blei

```

```python
m_begin = unc.ufloat(445.00, 0.05)
m_end = unc.ufloat(444.50, 0.05)
m_evap = m_begin - m_end
time = unc.ufloat(120, 3)

m_verdunstung = m_evap/time
m_verdunstung
```

```python
m1_graphit = unc.ufloat(444.4, 0.05)
m2_graphit = unc.ufloat(425.2, 0.05)
t_graphit = unc.ufloat(39, 3)

m1_alu = unc.ufloat(425.1, 0.05)
m2_alu = unc.ufloat(398.6, 0.05)
t_alu = unc.ufloat(105, 3)

m1_blei = unc.ufloat(398.5, 0.05)
m2_blei = unc.ufloat(379.9, 0.05)
t_blei = unc.ufloat(106, 3)

T_1 = unc.ufloat(24.10, 0.15)
T_2 = unc.ufloat(-195.8, 0.15)

Q_v = 199

M_graphit = 43.2
M_alu = 33.4
M_blei = 137.7


ttc_x = lambda m1, m2, t, Q_v, m_x, m_verdunstung, T_1, T_2: Q_v * (m1 - m2 - m_verdunstung*t) / (m_x * (T_1 - T_2))

ttc_graphit = ttc_x(m1_graphit, m2_graphit, t_graphit, Q_v, M_graphit, m_verdunstung, T_1, T_2)
ttc_alu = ttc_x(m1_alu, m2_alu, t_alu, Q_v, M_alu, m_verdunstung, T_1, T_2)
ttc_blei = ttc_x(m1_blei, m2_blei, t_blei, Q_v, M_blei, m_verdunstung, T_1, T_2)

f'{ttc_graphit=:.3uP}, {ttc_alu=:.3uP}, {ttc_blei=:.3uP}'
```
