```python
M_alu = unc.ufloat(155.70, 0.05)
M_graphit = unc.ufloat(125.1, 0.05)
M_blei = unc.ufloat(601.3, 0.05)

def c_x(T_bar, T1, T2, M_WK, M_K, M_x, c_w, W):
    return ((M_WK-M_K) * c_w + W) * (T_bar - T2) / (M_x * (T1 - T_bar))

c_graphit = c_x(T_bar_graphit, T1_graphit, T2_graphit, M_WK, M_K, M_graphit, c_w, W)
c_alu = c_x(T_bar_alu, T1_alu, T2_alu, M_WK, M_K, M_alu, c_w, W)
c_blei = c_x(T_bar_blei, T1_blei, T2_blei, M_WK, M_K, M_blei, c_w, W)
c_graphit, c_alu, c_blei

```
