```python
import numpy as np
```

```python
c0 = np.array([51545, 51483, 51625, 51615, 51549])
c1 = np.array([47577, 47315, 47219, 47096, 47206])
c2 = np.array([40052, 40003, 39741, 39913, 39952])
c3 = np.array([34170, 34188, 34145, 34061, 34315])
c4 = np.array([18601, 18644, 18496, 18573, 18646])

def calculate_mean(data):
    s = np.std(data, ddof=1)/np.sqrt(len(data))
    print(np.mean(data), s)
    
    
calculate_mean(c0)
calculate_mean(c1)
calculate_mean(c2)
calculate_mean(c3)
calculate_mean(c4)
```
