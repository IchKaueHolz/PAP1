Bessel Verfahren

```python
x1 = np.array([36.7, 36.60, 36.80])
x2 = np.array([65.5, 65.5, 65.4])
d = x2 - x1
std_d = np.std(d, ddof=1)
print(d)
L = 62.0
error_L = 0.15
error_d = np.sqrt(0.4**2 + std_d**2) #2.5mm pro ablesung

print(f"Mean d: {np.mean(d):.2f} ± {error_d:.2f}")
print(f"Mean L: {L:.2f} ± {error_L:.2f}")
```

Bessel-Verfahren mit Filter (blau)

```python

x1 = np.array([36.4, 36.4, 36.40])
x2 = np.array([65.4, 65.7, 65.7])
d = x2 - x1
std_d = np.std(d, ddof=1)
print(std_d)
L = 61.8
error_L = 0.10 # fucking messprotokoll hat aufeinamal 0.10 angegeben, obwohl es vorher 0.15 war
error_d = np.sqrt(0.4**2 + std_d**2)

print(f"Mean d: {np.mean(d):.2f} ± {error_d:.2f}")
print(f"Mean L: {L:.2f} ± {error_L:.2f}")
```

```python

L = 61.8
dL = 0.15

d_rot = 28.50
dd_rot = 0.44

d_blau = 29.20
dd_blau = 0.44

def calc_f(L, d):
    return (L**2 - d**2) / (4 * L)

def calc_df(L, dL, d, dd):
    term1 = ((L**2 + d**2) / (4 * L**2)) * dL
    term2 = (-d / (2 * L)) * dd
    return np.sqrt(term1**2 + term2**2)

f_rot = calc_f(L, d_rot)
df_rot = calc_df(L, dL, d_rot, dd_rot)

f_blau = calc_f(L, d_blau)
df_blau = calc_df(L, dL, d_blau, dd_blau)

print(f"Red: f = {f_rot:.4f} +/- {df_rot:.4f}")
print(f"Blue: f = {f_blau:.4f} +/- {df_blau:.4f}")
```
