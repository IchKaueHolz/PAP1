```python

L = 61.8
dL = 0.15

d_rot = 28.50
dd_rot = 0.44

d_blau = 29.20
dd_blau = 0.44

def calc_f(L, d):
    return (L**2 - d**2) / (4 * L)

def calc_df(L, dL, d, dd):
    term1 = ((L**2 + d**2) / (4 * L**2)) * dL
    term2 = (-d / (2 * L)) * dd
    return np.sqrt(term1**2 + term2**2)

f_rot = calc_f(L, d_rot)
df_rot = calc_df(L, dL, d_rot, dd_rot)

f_blau = calc_f(L, d_blau)
df_blau = calc_df(L, dL, d_blau, dd_blau)

print(f"Red: f = {f_rot:.4f} +/- {df_rot:.4f}")
print(f"Blue: f = {f_blau:.4f} +/- {df_blau:.4f}")
```

```python
# Brennweite Objektiv
f_1 = 40.0  # in mm (4 cm)

# Bildweite b und Fehler
b = 250.0  # in mm (25.0 cm)
db = 1.0   # in mm (0.1 cm)

# Zwischenbild Messstrecke und Fehler
B = 8.0    # in mm
dB = 0.3   # in mm
n = 16.0   # Anzahl der Striche

# Spaltbreiten für grünes Licht (aus den 3 Messungen)
D_messungen = np.array([0.200, 0.300, 0.300]) # in mm
# Abstand Spalt - Objekt für grünes Licht
s = 38.0   # in mm (laut Protokoll)
ds = 0.1   # in mm (angenommener Ablesefehler, anpassen falls nötig)

# Wellenlänge grünes Licht
lam = 550e-6 # in mm (550 nm)


# --- 1. Berechnung der Gitterkonstante ---

# Abbildungsmaßstab beta
beta = b / f_1 - 1
dbeta = db / f_1  # Fehlerfortpflanzung für beta

# Strichabstand im Zwischenbild
B_strich = B / n
dB_strich = dB / n

# Gitterkonstante des Objekts
G_strich = B_strich / beta
# Gaußsche Fehlerfortpflanzung für G_strich
dG_strich = G_strich * np.sqrt((dB_strich / B_strich)**2 + (dbeta / beta)**2)

print("--- 1. Gitterkonstante ---")
print(f"Abbildungsmaßstab beta = {beta:.3f} +/- {dbeta:.3f}")
print(f"Strichabstand Zwischenbild B_strich = {B_strich:.4f} +/- {dB_strich:.4f} mm")
print(f"Gitterkonstante G_strich = {G_strich*1000:.1f} +/- {dG_strich*1000:.1f} µm\n")


# --- 2. Theoretisches Auflösungsvermögen (Rayleigh) ---

# Mittelwert und Fehler der Spaltbreite D (Standardabweichung des Mittelwerts)
D_mean = np.mean(D_messungen)
dD_mean = np.std(D_messungen, ddof=1) / np.sqrt(len(D_messungen))

# Numerische Apertur NA (Näherung für kleine Winkel)
NA = D_mean / (2 * s)
# Fehler der NA (Gauß)
dNA = NA * np.sqrt((dD_mean / D_mean)**2 + (ds / s)**2)

# Theoretisches Auflösungsvermögen G_min
G_min_theo = 0.61 * lam / NA
# Fehler für G_min_theo
dG_min_theo = G_min_theo * (dNA / NA)

print("--- 2. Theoretisches Auflösungsvermögen ---")
print(f"Kritische Spaltbreite D (Grün) = {D_mean:.4f} +/- {dD_mean:.4f} mm")
print(f"Numerische Apertur NA = {NA:.5f} +/- {dNA:.5f}")
print(f"Theo. Auflösungsvermögen G_min_theo = {G_min_theo*1000:.1f} +/- {dG_min_theo*1000:.1f} µm")

print(f'\n\nich sitze hier im Audimax und verliere meinen\nVerstand als ich gesehen habe, dass ich noch mehr Fehlerfortplanzung\nmachen muss WTRFFTFWYT RUFSAF KSAFDR FHKJL ASFKLJ ASFHASF HSAFKLJA')

# das ist AI btw 
```
