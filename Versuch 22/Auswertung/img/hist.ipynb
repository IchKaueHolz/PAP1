```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np

# Q-Werte direkt in 10^-19 As
Q = np.array(
    [
        4.753,
        4.731,
        4.830,
        4.898,
        4.782,
        3.118,
        3.304,
        2.960,
        2.850,
        3.125,
        3.400,
        3.019,
        3.207,
        2.976,
        3.042,
        19.223,
        19.059,
        19.420,
        19.580,
        19.336,
        10.754,
        10.852,
        11.423,
        10.986,
        10.911,
        1.584,
        1.806,
        1.521,
        1.458,
        1.544,
    ]
)

bin_width = 0.2
bins = np.arange(0, Q.max() + bin_width, bin_width)

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(Q, bins=bins, color="steelblue", edgecolor="black")

# Markierungen bei Vielfachen der Elementarladung e ≈ 1.6 × 10^-19 As
E = 1.6
for n in range(1, int(np.ceil(Q.max() / E)) + 1):
    x = n * E
    ax.axvline(x, color="red", linestyle="--", linewidth=0.9, alpha=0.7)
    ax.text(
        x,
        ax.get_ylim()[1] * 0.98,
        rf"{n}e",
        color="red",
        fontsize=9,
        rotation=90,
        va="top",
        ha="right",
    )

# 1. Haupt-Ticks (Major Ticks) festlegen
ax.xaxis.set_major_locator(ticker.MultipleLocator(2.0))  # Alle 2 Einheiten
ax.yaxis.set_major_locator(ticker.MultipleLocator(1.0))  # Ganze Zahlen

# 2. Zwischen-Ticks (Minor Ticks) festlegen
ax.xaxis.set_minor_locator(
    ticker.MultipleLocator(0.5)
)  # Feine Unterteilung alle 0.5
ax.yaxis.set_minor_locator(ticker.MultipleLocator(0.5))

# 3. Optional: Aussehen der Ticks anpassen (Länge, Dicke, Richtung)
ax.tick_params(axis="both", which="major", length=6, width=1.2, direction="in")
ax.tick_params(axis="both", which="minor", length=3, width=0.8, direction="in")

ax.set_xlabel(r"$Q$ [$10^{-19}\ \mathrm{As}$]")
ax.set_ylabel("Häufigkeit")
ax.set_title(r"Histogramm der Ladung $Q$")

plt.tight_layout()
plt.show()
```
